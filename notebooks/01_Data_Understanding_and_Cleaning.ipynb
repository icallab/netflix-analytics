## DATA QUALITY ASSESSMENT

This section evaluates the quality and consistency of the Netflix dataset
by identifying missing values, duplicate records, invalid data,
and potential inconsistencies.

```python

# ==========================================
# MISSING VALUES ANALYSIS
# ==========================================

missing_analysis = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage (%)': (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_analysis = missing_analysis.sort_values(
    by='Missing Values',
    ascending=False
)

display(missing_analysis)
```

```python

# Missing Values Visualization

missing_data = missing_analysis[
    missing_analysis['Missing Values'] > 0
]

plt.figure(figsize=(12, 6))

sns.barplot(
    x=missing_data.index,
    y=missing_data['Percentage (%)']
)

plt.title('Missing Values Percentage by Column',
          fontsize=15, fontweight='bold')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# DUPLICATE DATA ANALYSIS
# ==========================================

# Check Full Row Duplicates
full_duplicates = df.duplicated().sum()

# Check Duplicate Show IDs
duplicate_ids = df['show_id'].duplicated().sum()

print("Duplicate Data Analysis")
print("=" * 40)
print(f"Full Duplicate Rows : {full_duplicates}")
print(f"Duplicate Show IDs  : {duplicate_ids}")
```

```python

# ==========================================
# DATA TYPE VALIDATION
# ==========================================

data_type_analysis = pd.DataFrame({
    'Column': df.columns,
    'Current Data Type': df.dtypes.astype(str).values
})

display(data_type_analysis)
```

```python

# ==========================================
# EMPTY STRING ANALYSIS
# ==========================================

empty_string_analysis = pd.DataFrame({
    'Empty Strings': [
        df[col].astype('string').str.strip().eq('').sum()
        for col in df.columns
    ]
}, index=df.columns)

display(
    empty_string_analysis.sort_values(
        by='Empty Strings',
        ascending=False
    )
)
```

```python

# ==========================================
# CATEGORICAL DATA VALIDATION
# ==========================================

# Content Type Validation
print("Content Type Distribution")
print("=" * 40)
print(df['type'].value_counts(dropna=False))

# Rating Validation
print("\nRating Distribution")
print("=" * 40)
print(df['rating'].value_counts(dropna=False))

# Release Year Validation
print("\nRelease Year Range")
print("=" * 40)
print(f"Minimum Release Year: {df['release_year'].min()}")
print(f"Maximum Release Year: {df['release_year'].max()}")

print("\nPotential Invalid Release Years:")
display(
    df[
        (df['release_year'] < 1900) |
        (df['release_year'] > pd.Timestamp.now().year)
    ][['show_id', 'title', 'release_year']]
)
```

```python

# ==========================================
# DATE VALIDATION
# ==========================================

date_check = pd.to_datetime(
    df['date_added'].str.strip(),
    format='%B %d, %Y',
    errors='coerce'
)

invalid_dates = df[
    df['date_added'].notna() &
    date_check.isna()
]

print("Date Validation")
print("=" * 40)
print(f"Missing Dates       : {df['date_added'].isna().sum()}")
print(f"Invalid Date Values : {len(invalid_dates)}")

display(
    invalid_dates[['show_id', 'title', 'date_added']].head(10)
)
```

```python

# ==========================================
# DURATION VALIDATION
# ==========================================

print("Duration Analysis")
print("=" * 40)

print("Duration by Content Type:")
display(
    df.groupby('type', dropna=False)['duration']
      .agg(['count', 'nunique'])
)

print("\nDuration Format Distribution:")

duration_format = pd.Series({
    'Movie Duration (min)': df['duration'].str.contains(
        r'^\d+\s+min$', na=False
    ).sum(),
    'TV Show Duration (Season)': df['duration'].str.contains(
        r'^\d+\s+Seasons?$', na=False
    ).sum(),
    'Missing Duration': df['duration'].isna().sum(),
    'Other Formats': (
        ~(
            df['duration'].str.contains(
                r'^\d+\s+min$|^\d+\s+Seasons?$',
                na=False
            ) |
            df['duration'].isna()
        )
    ).sum()
})

display(duration_format.to_frame('Total Records'))
```

```python

# ==========================================
# CONTENT TYPE & DURATION CONSISTENCY
# ==========================================

# Identify Movie Records with Season Format
movie_with_season = df[
    (df['type'] == 'Movie') &
    (df['duration'].str.contains(
        r'Season', case=False, na=False
    ))
]

# Identify TV Show Records with Minute Format
tv_with_minutes = df[
    (df['type'] == 'TV Show') &
    (df['duration'].str.contains(
        r'min', case=False, na=False
    ))
]

# Missing Duration by Content Type
missing_duration_by_type = (
    df[df['duration'].isna()]
    .groupby('type')
    .size()
)

print("Content Type Consistency")
print("=" * 40)

print(f"Movie with Season Format : {len(movie_with_season)}")
print(f"TV Show with Minute Format: {len(tv_with_minutes)}")

print("\nMissing Duration by Type:")
display(missing_duration_by_type)

print("\nPotential Inconsistent Records:")
display(pd.concat([
    movie_with_season,
    tv_with_minutes
])[['show_id', 'title', 'type', 'duration']])
```

```python

# ==========================================
# DATA QUALITY SUMMARY
# ==========================================

quality_summary = pd.DataFrame({
    'Quality Check': [
        'Total Records',
        'Total Columns',
        'Total Missing Values',
        'Rows with Missing Values',
        'Full Duplicate Rows',
        'Duplicate Show IDs',
        'Empty String Values'
    ],

    'Results': [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.isnull().any(axis=1).sum(),
        df.duplicated().sum(),
        df['show_id'].duplicated().sum(),
        sum(
            df[col].astype('string').str.strip().eq('').sum()
            for col in df.columns
        )
    ]
})

display(quality_summary)
```

## DATA CLEANING

```python
# Create a Copy of Original Dataset
df_clean = df.copy()

print("Original Dataset Shape:", df.shape)
print("Clean Dataset Shape   :", df_clean.shape)
```

```python

# ==========================================
# FIX RATING & DURATION INCONSISTENCY
# ==========================================

# Identify Incorrect Rating Values
incorrect_duration = ['66 min', '74 min', '84 min']

# Move Incorrect Rating Values to Duration
mask = df_clean['rating'].isin(incorrect_duration)

df_clean.loc[mask, 'duration'] = df_clean.loc[mask, 'rating']

# Set Incorrect Rating Values to Missing
df_clean.loc[mask, 'rating'] = np.nan

# Display Corrected Records
display(
    df_clean.loc[
        mask,
        ['show_id', 'title', 'type', 'rating', 'duration']
    ]
)

print("Corrected Records:", mask.sum())
```

```python

# ==========================================
# HANDLE MISSING VALUES
# ==========================================

# Columns to Fill with Unknown
columns_to_fill = [
    'director',
    'cast',
    'country',
    'rating'
]

# Replace Missing Values
for col in columns_to_fill:
    df_clean[col] = df_clean[col].fillna('Unknown')

# Check Missing Values After Treatment
display(
    df_clean[columns_to_fill].isnull().sum()
)
```

```python

# ==========================================
# DATE CONVERSION
# ==========================================

df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'].str.strip(),
    format='%B %d, %Y',
    errors='coerce'
)

# Verify Date Conversion
print("Date Column Data Type:")
print(df_clean['date_added'].dtype)

print("\nMissing Dates:")
print(df_clean['date_added'].isnull().sum())

display(df_clean[['title', 'date_added']].head())
```

```python

# ==========================================
# CLEANED DATASET VALIDATION
# ==========================================

print("CLEANED DATASET VALIDATION")
print("=" * 45)

print(f"Total Records       : {len(df_clean):,}")
print(f"Total Columns       : {len(df_clean.columns)}")
print(f"Duplicate Rows      : {df_clean.duplicated().sum()}")
print(f"Duplicate Show IDs  : {df_clean['show_id'].duplicated().sum()}")
print(f"Total Missing Values: {df_clean.isnull().sum().sum()}")

print("\nMissing Values by Column:")
display(
    df_clean.isnull().sum().to_frame('Missing Values')
)

print("\nData Types:")
display(
    df_clean.dtypes.to_frame('Data Type')
)
```

```python

# ==========================================
# BEFORE & AFTER COMPARISON
# ==========================================

cleaning_comparison = pd.DataFrame({
    'Before Cleaning': [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df['rating'].isin(incorrect_duration).sum()
    ],

    'After Cleaning': [
        len(df_clean),
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum(),
        df_clean['rating'].isin(incorrect_duration).sum()
    ]
}, index=[
    'Total Records',
    'Total Missing Values',
    'Duplicate Rows',
    'Incorrect Rating Values'
])

display(cleaning_comparison)
```
